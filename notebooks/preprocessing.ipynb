# Preprocessing

## Dim_Customers

```python
Dim_Customers.head()
```

```python
print("--- Data Info ---")
Dim_Customers.info()

print("\n--- Missing Values ---")
print(Dim_Customers.isnull().sum())

print("\n--- Duplicates ---")
print("Exact duplicates:", Dim_Customers.duplicated().sum())
print("Duplicate IDs:", Dim_Customers.duplicated(subset=['customer_unique_id']).sum())

# Descriptive statistics to detect any anomalies (e.g., out-of-bound coordinates)
Dim_Customers.describe()
```

```python
# 1. Initial Cleaning (As done previously)
Dim_Customers['zip_code'] = Dim_Customers['zip_code'].astype(str).str.zfill(5)
Dim_Customers['load_date_timestamp'] = pd.to_datetime(Dim_Customers['load_date_timestamp'])
Dim_Customers['city'] = Dim_Customers['city'].str.lower().str.strip()
Dim_Customers['state'] = Dim_Customers['state'].str.upper().str.strip()

# Set anomalous coordinates (outside Brazil) to NaN
lat_min, lat_max = -33.75, 5.27
lon_min, lon_max = -73.98, -34.79
invalid_mask = (
    (Dim_Customers['latitude'] < lat_min) | (Dim_Customers['latitude'] > lat_max) |
    (Dim_Customers['longitude'] < lon_min) | (Dim_Customers['longitude'] > lon_max)
)
Dim_Customers.loc[invalid_mask, ['latitude', 'longitude']] = np.nan

# 2. Prepare Data for Multiple Imputation
# We need to encode categorical strings into numbers for the model
encoder = OrdinalEncoder()
df_for_imputation = Dim_Customers.copy()

# Encode City and State
df_for_imputation[['city_encoded', 'state_encoded']] = encoder.fit_transform(df_for_imputation[['city', 'state']])

# Convert zip_code to float as it's a strong geographic indicator
df_for_imputation['zip_code_num'] = df_for_imputation['zip_code'].astype(float)

# Select the features that will participate in the Imputation algorithm
features_to_impute = ['zip_code_num', 'state_encoded', 'city_encoded', 'latitude', 'longitude']
data_matrix = df_for_imputation[features_to_impute]

print("Data matrix successfully prepared for Multiple Imputation...")
```

```python
# 3. Build and Run the Iterative Imputer (MICE)
# We use Random Forest for high accuracy (n_estimators=20 is used to balance speed and accuracy)
imputer = IterativeImputer(
    estimator=RandomForestRegressor(n_estimators=20, random_state=42),
    max_iter=10,        # Number of imputation iterations
    random_state=42,
    verbose=2           # To show progress during training
)

print("Running MICE Imputation (This may take a moment due to the size of the dataset)...")
# Apply the algorithm to fill the missing values
imputed_matrix = imputer.fit_transform(data_matrix)

# 4. Retrieve the imputed coordinates back into the original DataFrame
# Latitude and Longitude are at index 3 and 4 in the features_to_impute list
Dim_Customers['latitude'] = imputed_matrix[:, 3]
Dim_Customers['longitude'] = imputed_matrix[:, 4]

# Final check to ensure there are no missing values left
print("\nMissing values after Multiple Imputation:")
print(Dim_Customers.isnull().sum())
```

## Dim_Products

```python
# Display the first 5 rows to get an initial look
Dim_Products
```

```python
print("--- Data Info ---")
Dim_Products.info()

print("\n--- Missing Values ---")
print(Dim_Products.isnull().sum())

print("\n--- Impossible Physical Values ---")
print("Products with 0g weight:", (Dim_Products['weight_g'] == 0).sum())

# Let's check why there are more missing English categories than Portuguese ones
missing_en_but_has_pt = Dim_Products[Dim_Products['category_name_pt'].notnull() & Dim_Products['category_name_en'].isnull()]
print("\n--- Categories missing English translation ---")
print(missing_en_but_has_pt['category_name_pt'].unique())
```

```python
# 1. Fix Categorical Text Issues First
# Map known missing English translations
cat_mapping = {
    'pc gamer': 'pc gamer',
    'portateis cozinha e preparadores de alimentos': 'kitchen portables'
}
mask = Dim_Products['category_name_pt'].notnull() & Dim_Products['category_name_en'].isnull()
Dim_Products.loc[mask, 'category_name_en'] = Dim_Products.loc[mask, 'category_name_pt'].map(cat_mapping)

# Fill completely missing categories with 'unknown' so the model can group them together
Dim_Products['category_name_pt'] = Dim_Products['category_name_pt'].fillna('desconhecido')
Dim_Products['category_name_en'] = Dim_Products['category_name_en'].fillna('unknown')

# Fix impossible physical weights (0g) by turning them into NaN
Dim_Products['weight_g'] = Dim_Products['weight_g'].replace(0, np.nan)

# Format Date
Dim_Products['load_date_timestamp'] = pd.to_datetime(Dim_Products['load_date_timestamp'])

# 2. Prepare Data for KNN (Encoding & Feature Selection)
encoder = OrdinalEncoder()
df_knn = Dim_Products.copy()

# Encode the text categories
df_knn[['cat_pt_encoded', 'cat_en_encoded']] = encoder.fit_transform(df_knn[['category_name_pt', 'category_name_en']])

# Define the numerical matrix we want to impute
features_to_impute = [
    'cat_pt_encoded', 'cat_en_encoded', 
    'name_length', 'description_length', 'photo_count',
    'weight_g', 'length_cm', 'height_cm', 'width_cm'
]
data_matrix = df_knn[features_to_impute]

# 3. Feature Scaling (CRITICAL for KNN)
scaler = MinMaxScaler()
scaled_matrix = scaler.fit_transform(data_matrix)

print("Product data encoded and scaled. Ready for KNN Imputation...")
```

```python
# 4. Build and Run the KNN Imputer
knn_imputer = KNNImputer(n_neighbors=5, weights='distance')

print("Running KNN Imputation for Products...")
imputed_scaled_matrix = knn_imputer.fit_transform(scaled_matrix)

# 5. Inverse Transform (Un-scale back to real-world values)
imputed_matrix = scaler.inverse_transform(imputed_scaled_matrix)

# 6. Retrieve the imputed numerical columns back into the DataFrame
# The columns start from index 2 in our 'features_to_impute' list
cols_to_restore = ['name_length', 'description_length', 'photo_count', 'weight_g', 'length_cm', 'height_cm', 'width_cm']

for i, col in enumerate(cols_to_restore):
    Dim_Products[col] = imputed_matrix[:, i+2]  # +2 skips the encoded category columns

# 7. Post-Processing: Round discrete values to whole numbers
Dim_Products['name_length'] = Dim_Products['name_length'].round()
Dim_Products['description_length'] = Dim_Products['description_length'].round()
Dim_Products['photo_count'] = Dim_Products['photo_count'].round()

# 8. Recalculate Volume using the newly imputed & accurate dimensions
Dim_Products['volume_cm3'] = Dim_Products['length_cm'] * Dim_Products['height_cm'] * Dim_Products['width_cm']

# Final Check
print("\nMissing values after KNN Imputation:")
print(Dim_Products.isnull().sum())
print("\nProducts with 0g weight:", (Dim_Products['weight_g'] == 0).sum())
```

```python
# Display initial information
print("--- Initial Sellers Info ---")
Dim_Sellers.info()
print("\n--- Missing Values ---")
print(Dim_Sellers.isnull().sum())
```

```python
# 1. Fix Zip Codes (Pad with leading zeros to ensure 5 digits)
Dim_Sellers['zip_code'] = Dim_Sellers['zip_code'].astype(str).str.zfill(5)

# 2. Handle missing text/categorical data (Without dropping any columns)
# Fill missing cities with 'unknown'
Dim_Sellers['city'] = Dim_Sellers['city'].fillna('unknown').str.lower().str.strip()
Dim_Sellers['state'] = Dim_Sellers['state'].str.upper().str.strip()

# Fill missing marketing columns with 'unknown' as requested
marketing_cols = ['business_segment', 'business_type', 'lead_type', 'behaviour_profile']
for col in marketing_cols:
    Dim_Sellers[col] = Dim_Sellers[col].fillna('unknown').str.lower().str.strip()

# 3. Format the timestamp
Dim_Sellers['load_date_timestamp'] = pd.to_datetime(Dim_Sellers['load_date_timestamp'])

# 4. Handle out-of-bounds coordinates (outside Brazil) by setting them to NaN for Imputation
lat_min, lat_max = -33.75, 5.27
lon_min, lon_max = -73.98, -34.79

invalid_mask = (
    (Dim_Sellers['latitude'] < lat_min) | (Dim_Sellers['latitude'] > lat_max) |
    (Dim_Sellers['longitude'] < lon_min) | (Dim_Sellers['longitude'] > lon_max)
)
Dim_Sellers.loc[invalid_mask, ['latitude', 'longitude']] = np.nan

print("Initial preprocessing complete. Total columns maintained:", len(Dim_Sellers.columns))
```

```python
# 1. Prepare Data and encode text so the KNN model can process it
encoder = OrdinalEncoder()
df_knn = Dim_Sellers.copy()

# Encode City and State into numbers
df_knn[['city_encoded', 'state_encoded']] = encoder.fit_transform(df_knn[['city', 'state']])
df_knn['zip_code_num'] = df_knn['zip_code'].astype(float)

# Select the features that will participate in the KNN study
features_to_impute = ['state_encoded', 'city_encoded', 'zip_code_num', 'latitude', 'longitude']
data_matrix = df_knn[features_to_impute]

# 2. Feature Scaling (Crucial for KNN to balance calculations)
scaler = MinMaxScaler()
scaled_matrix = scaler.fit_transform(data_matrix)

# 3. Build and Run the KNN Imputer
print("Running KNN Imputation to estimate missing coordinates...")
knn_imputer = KNNImputer(n_neighbors=5, weights='distance')
imputed_scaled_matrix = knn_imputer.fit_transform(scaled_matrix)

# 4. Inverse Transform (Un-scale back to real GPS coordinates)
imputed_matrix = scaler.inverse_transform(imputed_scaled_matrix)

# 5. Retrieve the newly imputed coordinates back into the original DataFrame
# Latitude and Longitude are at index 3 and 4 in our features_to_impute list
Dim_Sellers['latitude'] = imputed_matrix[:, 3]
Dim_Sellers['longitude'] = imputed_matrix[:, 4]

# Display the result to verify no missing values remain in the processed columns
print("\nMissing values in processed columns after Imputation:")
print(Dim_Sellers[['city', 'state', 'latitude', 'longitude', 'business_segment']].isnull().sum())
```
